```python
# Uploading/Looking into Dataset block
import pandas as pd
import gdown
url = "https://drive.google.com/file/d/1-bC4EMNEHbu8kYWwPwhiU4MOzapEav0s/view?usp=sharing"
filename = "csb.csv"
gdown.download(url, filename, fuzzy=True)
df = pd.read_csv(filename)
df.head(10)
```

```python
df.describe(include = 'all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating']. transform(lambda x: x.fillna(x.median()))
df.isnull().sum()
```

```python
df.columns
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
df['age'].unique()
```

```python
labels = ['Young Adult','Adult','Middle aged','Senior']
df['age_group'] = pd.cut(df['age'], bins=[0,30,50,60,100],labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
frequency_mapping = {
    'Fortnightly' :14,
    'Bi-Weekly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Annually': 365
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['frequency_of_purchases','purchase_frequency_days']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

```python
df.to_csv('csb.csv', index=False)
```
